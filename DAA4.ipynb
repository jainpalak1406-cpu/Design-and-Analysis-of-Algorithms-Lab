```python
import random
import time
import matplotlib.pyplot as plt
import numpy as np

# SELECTION AND RECURSIVE TECHNIQUES

# Maximum product of array elements
def maximize_product(arr):

    arr.sort()

    product = 1

    for i in range(len(arr) - 1, -1, -1):

        if arr[i] > 0:
            product *= arr[i]

    return product


sizes = [100, 500, 1000, 2000, 5000, 10000]
times = []

for n in sizes:

    arr = [random.randint(1, 100) for _ in range(n)]

    start = time.perf_counter()
    maximize_product(arr)
    end = time.perf_counter()

    times.append(end - start)

plt.figure()
plt.plot(sizes, times, marker='o')
plt.xlabel("Input size (n)")
plt.ylabel("Execution time (seconds)")
plt.title("Maximum Product of Array Elements")
plt.grid(True)
plt.show()



# Quickselect kth largest
def partition(arr, low, high):

    pivot = arr[high]
    i = low

    for j in range(low, high):

        if arr[j] > pivot:

            arr[i], arr[j] = arr[j], arr[i]
            i += 1

    arr[i], arr[high] = arr[high], arr[i]

    return i


def kth_largest(arr, k):

    low = 0
    high = len(arr) - 1

    while low <= high:

        p = partition(arr, low, high)

        if p == k - 1:
            return arr[p]

        elif p > k - 1:
            high = p - 1

        else:
            low = p + 1


sizes = [1000, 2000, 4000, 8000, 16000]
times = []

for n in sizes:

    arr = random.sample(range(n * 2), n)

    k = n // 2

    start = time.perf_counter()
    kth_largest(arr.copy(), k)
    end = time.perf_counter()

    times.append(end - start)

plt.figure()
plt.plot(sizes, times, marker='o')
plt.xlabel("Input size (n)")
plt.ylabel("Execution time (seconds)")
plt.title("Quickselect - Kth Largest Element")
plt.grid(True)
plt.show()




# Fibonacci - Recursion
def fibonacci_recursive(n):

    if n <= 1:
        return n

    return (
        fibonacci_recursive(n - 1)
        + fibonacci_recursive(n - 2)
    )


sizes = list(range(5, 31, 2))
times = []

for n in sizes:

    start = time.perf_counter()
    fibonacci_recursive(n)
    end = time.perf_counter()

    times.append(end - start)

plt.figure()
plt.plot(sizes, times, marker='o')
plt.xlabel("n")
plt.ylabel("Execution time (seconds)")
plt.title("Fibonacci Using Recursion")
plt.grid(True)
plt.show()



# Fibonacci - Dynamic Programming
def fibonacci_dp(n):

    if n <= 1:
        return n

    dp = [0] * (n + 1)

    dp[0] = 0
    dp[1] = 1

    for i in range(2, n + 1):
        dp[i] = dp[i - 1] + dp[i - 2]

    return dp[n]


sizes = [1000, 2000, 4000, 8000, 16000, 32000]
times = []

for n in sizes:

    start = time.perf_counter()
    fibonacci_dp(n)
    end = time.perf_counter()

    times.append(end - start)

plt.figure()
plt.plot(sizes, times, marker='o')
plt.xlabel("n")
plt.ylabel("Execution time (seconds)")
plt.title("Fibonacci Using Dynamic Programming")
plt.grid(True)
plt.show()



# Smallest missing positive integer
def smallest_missing(arr):

    arr = set(arr)

    i = 1

    while i in arr:
        i += 1

    return i


sizes = [1000, 2000, 4000, 8000, 16000, 32000]
times = []

for n in sizes:

    arr = list(range(1, n + 1))
    arr[random.randint(0, n - 1)] = -1

    start = time.perf_counter()
    smallest_missing(arr)
    end = time.perf_counter()

    times.append(end - start)

plt.figure()
plt.plot(sizes, times, marker='o')
plt.xlabel("Input size (n)")
plt.ylabel("Execution time (seconds)")
plt.title("Smallest Missing Positive Integer")
plt.grid(True)
plt.show()

```
