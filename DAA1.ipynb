```python
import random
import time
import matplotlib.pyplot as plt
import numpy as np

# SORTING FUNDAMENTALS

# Iterative insertion sort
def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1

        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1

        arr[j + 1] = key

    return arr


sizes = [100, 200, 400, 800, 1600]
times = []

for n in sizes:
    arr = random.sample(range(n * 2), n)

    start = time.perf_counter()
    insertion_sort(arr.copy())
    end = time.perf_counter()

    times.append(end - start)

plt.figure()
plt.plot(sizes, times, marker='o')
plt.xlabel("Input size (n)")
plt.ylabel("Execution time (seconds)")
plt.title("Insertion Sort - Iterative")
plt.grid(True)
plt.show()


# Recursive insertion sort
def insertion_sort_recursive(arr, n):
    if n <= 1:
        return

    insertion_sort_recursive(arr, n - 1)

    key = arr[n - 1]
    j = n - 2

    while j >= 0 and arr[j] > key:
        arr[j + 1] = arr[j]
        j -= 1

    arr[j + 1] = key


sizes = [100, 200, 400, 800]
times = []

for n in sizes:
    arr = random.sample(range(n * 2), n)

    start = time.perf_counter()
    insertion_sort_recursive(arr.copy(), n)
    end = time.perf_counter()

    times.append(end - start)

plt.figure()
plt.plot(sizes, times, marker='o')
plt.xlabel("Input size (n)")
plt.ylabel("Execution time (seconds)")
plt.title("Insertion Sort - Recursive")
plt.grid(True)
plt.show()


# Dutch national flag problem
def sort_012(arr):
    low = 0
    mid = 0
    high = len(arr) - 1

    while mid <= high:

        if arr[mid] == 0:
            arr[low], arr[mid] = arr[mid], arr[low]
            low += 1
            mid += 1

        elif arr[mid] == 1:
            mid += 1

        else:
            arr[mid], arr[high] = arr[high], arr[mid]
            high -= 1

    return arr


sizes = [1000, 2000, 4000, 8000, 16000, 32000]
times = []

for n in sizes:
    arr = [random.randint(0, 2) for _ in range(n)]

    start = time.perf_counter()
    sort_012(arr.copy())
    end = time.perf_counter()

    times.append(end - start)

plt.figure()
plt.plot(sizes, times, marker='o')
plt.xlabel("Input size (n)")
plt.ylabel("Execution time (seconds)")
plt.title("Dutch National Flag Algorithm")
plt.grid(True)
plt.show()


# Majority element in an array
def majority_element(arr):
    candidate = None
    count = 0

    for num in arr:

        if count == 0:
            candidate = num

        if num == candidate:
            count += 1
        else:
            count -= 1

    if arr.count(candidate) > len(arr) // 2:
        return candidate

    return None


sizes = [1000, 2000, 4000, 8000, 16000, 32000]
times = []

for n in sizes:

    arr = [1] * (n // 2 + 1)
    arr += [random.randint(2, 100) for _ in range(n - len(arr))]
    random.shuffle(arr)

    start = time.perf_counter()
    majority_element(arr)
    end = time.perf_counter()

    times.append(end - start)

plt.figure()
plt.plot(sizes, times, marker='o')
plt.xlabel("Input size (n)")
plt.ylabel("Execution time (seconds)")
plt.title("Majority Element - Boyer Moore")
plt.grid(True)
plt.show()



```
